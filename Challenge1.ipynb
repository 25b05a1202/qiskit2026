```python

from qiskit import QuantumCircuit, transpile
from qiskit_aer import AerSimulator
from qiskit.visualization import plot_histogram
import matplotlib.pyplot as plt

simulator = AerSimulator()
```

```python

qc_a = QuantumCircuit(1, 1)

qc_a.x(0)
qc_a.measure(0, 0)

qc_a.draw("mpl")
```

```python

job_a = simulator.run(transpile(qc_a, simulator), shots=100)
counts_a = job_a.result().get_counts()

print("Circuit A Results:", counts_a)

plot_histogram(counts_a)
plt.show()
```

```python

qc_b = QuantumCircuit(1, 1)

qc_b.x(0)
qc_b.x(0)
qc_b.measure(0, 0)

qc_b.draw("mpl")
```

```python

job_b = simulator.run(transpile(qc_b, simulator), shots=100)
counts_b = job_b.result().get_counts()

print("Circuit B Results:", counts_b)

plot_histogram(counts_b)
plt.show()
```

```python

qc_c = QuantumCircuit(1, 1)

qc_c.h(0)
qc_c.measure(0, 0)

qc_c.draw("mpl")
```

```python

job_c = simulator.run(transpile(qc_c, simulator), shots=100)
counts_c = job_c.result().get_counts()

print("Circuit C Results:", counts_c)

plot_histogram(counts_c)
plt.show()
```

```python

print("Circuit A (One X Gate):", counts_a)
print("Circuit B (Two X Gates):", counts_b)
print("Circuit C (One H Gate):", counts_c)
```
