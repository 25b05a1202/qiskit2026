```python
from qiskit import QuantumCircuit, transpile
from qiskit_aer import AerSimulator
from qiskit.visualization import plot_histogram
import matplotlib.pyplot as plt

simulator = AerSimulator()
```

```python
qc_a = QuantumCircuit(2, 2)

qc_a.h(0)
qc_a.cx(0, 1)

qc_a.measure([0, 1], [0, 1])

qc_a.draw("mpl")
```

```python
job_a = simulator.run(
    transpile(qc_a, simulator),
    shots=100
)

counts_a = job_a.result().get_counts()

print("Part A Results:", counts_a)

plot_histogram(counts_a)
plt.show()
```

```python
qc_b = QuantumCircuit(2, 2)

qc_b.h(0)
qc_b.x(0)
qc_b.cx(0, 1)

qc_b.measure([0, 1], [0, 1])

qc_b.draw("mpl")
```

```python
job_b = simulator.run(
    transpile(qc_b, simulator),
    shots=100
)

counts_b = job_b.result().get_counts()

print("Part B Results:", counts_b)

plot_histogram(counts_b)
plt.show()
```

```python
print("Part A - Correlated:", counts_a)
print("Part B - Opposite:", counts_b)

plot_histogram(
    [counts_a, counts_b],
    legend=["Part A - Correlated", "Part B - Opposite"]
)
plt.show()
```
