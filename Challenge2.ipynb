```python
from qiskit import QuantumCircuit, transpile
from qiskit_aer import AerSimulator
from qiskit.visualization import plot_histogram
from math import pi
import matplotlib.pyplot as plt

simulator = AerSimulator()
```

```python
qc_a = QuantumCircuit(1, 1)

qc_a.ry(pi/3, 0)
qc_a.measure(0, 0)

qc_a.draw("mpl")
```

```python
job_a = simulator.run(
    transpile(qc_a, simulator),
    shots=100
)

counts_a = job_a.result().get_counts()

print("Circuit A Results:", counts_a)

plot_histogram(counts_a)
plt.show()
```

```python
qc_b = QuantumCircuit(1, 1)

qc_b.ry(2*pi/3, 0)
qc_b.measure(0, 0)

qc_b.draw("mpl")
```

```python
job_b = simulator.run(
    transpile(qc_b, simulator),
    shots=100
)

counts_b = job_b.result().get_counts()

print("Circuit B Results:", counts_b)

plot_histogram(counts_b)
plt.show()
```

```python
print("Circuit A (0 occurs more):", counts_a)
print("Circuit B (1 occurs more):", counts_b)
```
